# Testing Correctness

```python
def DeltaP2(Delta_p):
    dim = len(Delta_p)
    
    output = np.empty(dim*(dim+1)//2, dtype=float)
    
    idx=0
    
    for i in range(dim):
        for j in range(i, dim):
            output[idx] = Delta_p[i]*Delta_p[j]
            idx+=1
    return output

def DeltaP3(Delta_p):
    dim = len(Delta_p)
    
    output = np.empty(dim*(dim+1)*(dim+2)//6, dtype=float)
    
    idx = 0
    for i in range(dim):
        for j in range(i, dim):
            for k in range(j, dim):
                output[idx] = Delta_p[i]*Delta_p[j]*Delta_p[k]
                idx += 1
    return output

def DeltaP4(Delta_p):
    
    dim = len(Delta_p)
    
    output = np.empty(dim*(dim+1)*(dim+2)*(dim+3)//24, dtype=float)
    
    idx = 0
    for i in range(dim):
        for j in range(i, dim):
            for k in range(j, dim):
                for l in range(k, dim):
                    output[idx] = (
                        Delta_p[i]*Delta_p[j]*Delta_p[k]*Delta_p[l]
                    )
                    idx += 1
    return output


def DeltaP5(Delta_p):
    dim = len(Delta_p)
    
    output = np.empty(dim*(dim+1)*(dim+2)*(dim+3)*(dim+4)//120, dtype=float)
    
    idx = 0
    for i in range(dim):
        for j in range(i, dim):
            for k in range(j, dim):
                for l in range(k, dim):
                    for m in range(l, dim):
                        output[idx] = (
                            Delta_p[i]*Delta_p[j]*Delta_p[k]
                            *Delta_p[l]*Delta_p[m]
                        )
                        idx += 1
    return output


def DeltaP6(Delta_p):
    dim = len(Delta_p)
    
    output = np.empty(dim*(dim+1)*(dim+2)*(dim+3)*(dim+4)*(dim+5)//720, dtype=float)
    
    idx = 0
    for i in range(dim):
        for j in range(i, dim):
            for k in range(j, dim):
                for l in range(k, dim):
                    for m in range(l, dim):
                        for n in range(m, dim):
                            output[idx] = (
                                Delta_p[i]*Delta_p[j]*Delta_p[k]
                                *Delta_p[l]*Delta_p[m]*Delta_p[n]
                            )
                            idx += 1
    return output
```

```python
def test_f(delta_P):
    P2 = DeltaP2(delta_P)
    P3 = DeltaP3(delta_P)
    P4 = DeltaP4(delta_P)
    P5 = DeltaP5(delta_P)
    P6 = DeltaP6(delta_P)

    return np.concatenate((
        P2,P3,P4,P5,P6
    ))
```

```python
test_f(v) - Symmetric.Taylor_DeltaPs(v, 3)
```

```python
Symmetric.Taylor_DeltaPs(v, 3)
```

```python
def random_test():
    v = np.random.rand(12)
    res1 = Symmetric.Taylor_DeltaPs(v, 3)
    res2 = test_f(v)
    res = np.abs(res1-res2)

    return res.min(), res.max()
```

```python
random_test()
```
